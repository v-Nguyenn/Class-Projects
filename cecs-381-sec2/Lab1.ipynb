```python
# Part - 1: Define a function (self - implemented) to generate permutations
def gen_perms(elements):
    results = [] # initialize an empty return list
    
    # define the internal backtracking function
    def backtrack(current_path, remaining_options):
        # 1. Base case: no more options left to pick
        if len(remaining_options) == 0:
            results.append(current_path)
            return
        
        # 2. Recursive step: iterate thru all options
        for i in range(len(remaining_options)):
            
            # 2.1 Pick the item
            item = remaining_options[i]
            
            # 2.2 Form the new path (add the item to current path)
            new_path = current_path + [item]
            
            # 2.3 Form the new options (remove item out of remaining)
            new_options = remaining_options[:i] + remaining_options[i+1:]
            
            # 2.4 Recursive call
            backtrack(new_path, new_options)
        
    # Initial call
    backtrack([], elements)
    return results
```

```python
# Test Demo
input_set = ['A', 'B', 'C']
res1 = gen_perms(input_set)
for p in res1:
    print(p)
```

```python
# Part 1B: Use python library to implement permutation
import itertools
# 1. Create the iterator: this doesn't generate all data immediately, 
#    It prepares to generate them on the fly
perms_iterator = itertools.permutations(input_set)
print(perms_iterator) # You will only get a memory address here so we have to convert it 

# 2. Consuming the iterator (convert to a list)
perm_list = list(perms_iterator)
for p in perm_list:
    print(p)
```

```python
# Part 2A: Self implement combinations
def gen_combs(elements, k):
    results = []
    def backtrack(start_index, current_path):
        # 1. Base case: combination is full
        if len(current_path) == k:
            results.append(current_path[:]) # copy the list
            return
        # 2. Recursive step: we only iterate from start_index to the end
        # Note: this prevents duplicates
        for i in range(start_index, len(elements)):
            item = elements[i]
            # 2.1 Choose
            current_path.append(item)
            # 2.2 Recurse
            backtrack(i + 1, current_path)
            # 2.3 Undo the choice
            current_path.pop()

    backtrack(0, [])
    return results
```

```python
# test demo
data = ['A', 'B', 'C', 'D']
k = 2
combos = gen_combs(data, k)
for c in combos:
    print(c)
```

```python
# Part 2B: use lib to do combinations
combo_iterator = itertools.combinations(data, k)
combo_list = list(combo_iterator) # converting

for c in combo_list:
    print(c)
```
